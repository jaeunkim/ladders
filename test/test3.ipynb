## Testing the commutator() method
From Crimin 2018

```python
import sys
sys.path.append("..")
from ladders import Expression, commutator, print_nonzero_terms
```

```python
Lz = Expression("x+_y(+)-1x_y+")
x_annihilation = Expression("x")
x_creation = Expression("x+")
x_number = Expression("x+_x")
y_number = Expression("y+_y")
```

```python
print_nonzero_terms(commutator(Lz, x_annihilation))
print_nonzero_terms(commutator(Lz, x_creation))
print_nonzero_terms(commutator(Lz, x_number))
print_nonzero_terms(commutator(Lz, y_number))
```

```python
assert(commutator(Lz, x_annihilation).expr_dict["y"] == -1)
assert(commutator(Lz, x_creation).expr_dict["y+"] == -1)

assert(commutator(Lz, x_number).expr_dict["x+_y"] == -1)
assert(commutator(Lz, x_number).expr_dict["x_y+"] == -1)
```

```python
# verified with hand calculation
nx_first_commmutator = commutator(Lz, x_number)
nx_second_commmutator = commutator(Lz, nx_first_commmutator)
print_nonzero_terms(nx_second_commmutator)
```

```python
assert(nx_second_commmutator.expr_dict["x+_x"] == -2)
assert(nx_second_commmutator.expr_dict["y+_y"] == 2)
```

```python
ny_first_commmutator = commutator(Lz, y_number)
ny_second_commmutator = commutator(Lz, ny_first_commmutator)
print_nonzero_terms(ny_second_commmutator)
```
